```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_wine

data = load_wine(as_frame=True)
df = data.frame.copy()
display(df.head())
print(df.describe().T.head())

fig, ax = plt.subplots()
ax.hist(df['alcohol'], bins=20)
ax.set_title('Distribusi Alcohol')
ax.set_xlabel('Alcohol')
ax.set_ylabel('Frekuensi')
plt.show()

fig, ax = plt.subplots()
ax.scatter(df['alcohol'], df['color_intensity'], c=df['target'])
ax.set_xlabel('Alcohol')
ax.set_ylabel('Color intensity')
ax.set_title('Alcohol vs Color Intensity')
plt.show()
```

## Latihan Terstruktur
- Rumuskan tiga pertanyaan EDA dan pilih visualisasi untuk menjawabnya.
- Temukan minimal dua pola atau anomali dari data.

```python
# TODO: Kerjakan latihan pada sel ini dan tambahkan sel baru bila diperlukan.

#1. bagaimana distribusi kadar senyawa flavanoids pada sampel wine?
fig, ax = plt.subplots()
ax.hist(df['flavanoids'], color='blue', edgecolor='black', bins=20)
ax.set_title('distribusi kadar flavanoids pada wine')
ax.set_xlabel('kadar flavanoids')
ax.set_ylabel('jumlah sampel')
plt.show()

#2. apakah ada hubungan antar kadar plorin dan alkohol dalam wine?
fig, ax = plt.subplots()
ax.scatter(df['alcohol'], df['proline'], c=df['target'], cmap='viridis')
ax.set_xlabel('proline')
ax.set_ylabel('alcohol')
ax.set_title('hubungan kadar proline vs alkohol')
plt.show()

#3. apakah ada outlier?
fig, ax = plt.subplots()
ax.boxplot(df['color_intensity'])
ax.set_ylabel('intensitas warna')
ax.set_title('deteksi outlier pada intensitas warna wine')
plt.show()


```
