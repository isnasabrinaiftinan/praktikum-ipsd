e. validasi inputnya pakai len(df) == 0, ini maksudnya kalau tabelnya kosong maka fungsi berhenti, selain itu juga mengecek 'total_harga' dengan mengalikan 'jumlah_barang' dengan 'harga_barang'.

alasan pakai pandas = karena lebih ringkas untuk buat fungsi rata-rata [.mean()], jumlah[.sum()], dan modus[.mode()]

```python
import pandas as pd
import numpy as np

transaksi = [
    {'id': 1, 'kategori': 'A', 'qty': 2, 'harga': 15000},
    {'id': 2, 'kategori': 'B', 'qty': 1, 'harga': 25000},
    {'id': 3, 'kategori': 'A', 'qty': 4, 'harga': 12000},
    {'id': 4, 'kategori': 'C', 'qty': 3, 'harga': 18000},
]
df = pd.DataFrame(transaksi)
df['nilai'] = df['qty'] * df['harga']
display(df)
print('Total nilai:', df['nilai'].sum())
print('Rata-rata:', df['nilai'].mean())
```

```python
def ringkas_transaksi(df):

    if len(df) == 0:
        return "Data kosong"

    if 'harga' not in df.columns or 'qty' not in df.columns:
        return "ada kolom yang kosong"

    df['nilai'] = df['qty'] * df['harga']


    jumlah_baris = len(df)
    total_nilai = df['nilai'].sum()
    rata_rata_nilai = df['nilai'].mean()
    kategori_terbanyak = df['kategori'].mode()[0]

    return jumlah_baris, total_nilai, rata_rata_nilai, kategori_terbanyak

jml, total, avg, modus = ringkas_transaksi(df)

print("Jumlah baris       :", jml)
print("Total nilai        :", total)
print("Rata-rata          :", avg)
print("Kategori terbanyak :", modus)
```
